1. MSE 

Mean Squared Error (MSE): <br>


$$ C(\Theta) = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \tilde{y}_i)^2$$

where: 
n: number of training samples. <br>
$y_i$: The true target value for sample i. <br>
$ \tilde{y}_i$: The network's predicted output for sample i (the activation of the final layer). <br>

MSE with L1 and L2 norms. <br>

L2 norms (Ridge): <br>

$$
C_{L2}(\Theta) = C(\Theta) + \lambda \sum_{i,j} w_{ij}^2
$$

L2 norms adds a term proportional to the sum of the squares of the weights. <br>

L1 norms (lasso): <br>

The L1 norm uses the sum of the absolute values of the weights. 

$$
C_{L2}(\Theta) = C(\Theta) + \lambda \sum_{i,j} | w_{ij}|
$$
